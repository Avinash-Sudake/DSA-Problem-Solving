```python
 # List Comprehension

v = ["hello","python","Data Science","Java"]
z=[len(i)for i in v]
print(z)
```

```python
# Write a program to convert eatch string in list into uppercase

v = ["hello","python","Data Science","Java"]
z=[i.upper()for i in v]
print(z)
```

```python
# Write a program to convert eatch string in list into lowercase

v = ["hello","python","Data Science","Java"]
z=[i.lower()for i in v]
print(z)
```

```python
# Find a square of eatch number in a list with the help if list Comprehension

v = [2,4,6,8]
z=[i**2 for i in v]
print(z)
```

```python
# write a program taking a list of numbers from user and convert into int using list Comprehension

a=input("Enter a list of numbers  ")
z=[int(num) for num in a.split()]
print(z)
```

```python
# if statement in list Comprehension

age = [20,18,13,16,32,25,21]
g=[i for i in age if i>=18]
print(g)
```

```python
# Write a program to find a even num with the help of list Comprehension

num = [20,18,13,16,32,25,21]
g=[i for i in num if i%2==0]
print("Even Numbers :",g)
```

```python
# Write a program to find vowels of string with help of list Comprehension

j="Data Science"
vowels = [i for i in j if i in "aeiouAEIOU"]
print("Vowels:", vowels)
```

```python
# Write a program to find commoun element from two list

a=[2,4,6,8]
b=[4,6,12,14]
c=[i for i in a if i in b]
print("Common Elements :",c)
```

```python
# Write a program to find odd and even numbers in list Comprehension
num = [1,2,3,4,6,8]
g=["even" if i%2==0 else "odd" for i in num]
print(g)
```
