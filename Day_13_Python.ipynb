```python
# Exception Handling

for i in range(5,-5,-1):
  try:
    print(10/i)
  except:
    print("Logical Error")
```

```python
try:
  print(int("Hello"))
except:
        print("plz check your code")
```

```python
from typing_extensions import final
try:
  print(15/0)
except ValueError:
  print("This is value error")
except ZeroDivisionError:
  print("This is ZeroDivision error")
finally:
  print("Try block 1")
```

```python
# Class & Object

class aadhar:
  Name = "Avi Sudake"
  DOB = "05/07/2004"
  Address = "Pune"

xerox1=aadhar()
print("Xerox 1")
print(xerox1.Name)
print(xerox1.DOB)
print(xerox1.Address)
```

```python
class car:
  cname = "Tata"
  def __init__(self,mname):
    self.car_name=mname
c1=car("Nexon")
print(c1.cname)
print(c1.car_name)
c2=car("Punch")
print(c2.cname)
print(c2.car_name)
```
