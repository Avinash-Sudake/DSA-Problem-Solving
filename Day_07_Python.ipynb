```python
# Write A program To Find The Sum Of Didit Of a Given Number
g = input ("Enter The Number :")
g=g.split()
sumt = 0
for i in g :
  sumt = sumt + int(i)
print(sumt)
```

```python
# Print 5 table using while loop
num= int(input("Enter The Number"))
i=1
while i<11:
  print(f"{num} X {i} = {num*i}")
  i=i+1
```

```python
# Create a menu system taking multipal orders and print total bill after the order is completed


print("---------- हॉटेल MCA चायवाला ----------")
print("_______________________________________")
print("Menu")
print("1. Vadapav - ₹20")
print("2. Tea      - ₹10")
print("3. Samosa   - ₹30")
print("4. Bhel     - ₹40")
print("_______________________________________")

total_bill = 0

while True:
    order = int(input("Enter Your Order (1-4): "))

    if order == 1:
        print("Vadapav Added - ₹20")
        total_bill += 20
    elif order == 2:
        print("Tea Added - ₹10")
        total_bill += 10
    elif order == 3:
        print("Samosa Added - ₹30")
        total_bill += 30
    elif order == 4:
        print("Bhel Added - ₹40")
        total_bill += 40
    else:
        print("Sorry... This Menu Is Not Available")

    choice = input("Do you want to order more? (y/n): ").lower()

    if choice != "y":
        break

print("\n========== BILL ==========")
print("Total Bill = ₹", total_bill)
print("Thank You! Visit Again 😊")
```
