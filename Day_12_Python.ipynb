```python
# Write a program to find the length of eatch element in a list with the help of Lambda function

b=["hello","world","python"]
v=lambda c: [len(i) for i in c ]
print(v(b))
```

```python
# Write a program to find even and odd number from a list  with the help of lambda fun

b=[1,2,3,4,5,6,7]
v=lambda c: [f"Even:{i}"if i%2==0 else f"Odd:{i}" for i in c ]
print(v(b))
```

```python
# Exception Handling

for i in range(5,-5,-1):
  try:
    print(10/i)
  except:
    print("Logical Error")
```
