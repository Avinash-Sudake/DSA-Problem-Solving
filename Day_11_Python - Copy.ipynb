```python
# Write a program to check Armstrong number or not

num=int(input("enter Number :"))
l=len(str(num))
sum=0
temp=num
while temp>0:
  digit=temp%10
  sum=sum+digit**l
  temp=temp//10

if num == sum:
  print(f"{num} is an Armstrong number")
else:
  print(f"{num} is not an Armstrong number")
```

```python
# Write a program to reverse a num without using string reversing fun

num=int(input("Enter A Number :"))

rev = 0
temp = num

while num > 0:
  digit = num % 10
  rev = rev * 10 + digit
  num = num // 10

print(f"The reversed number is : {rev}")
```

```python
# Arbitary Argument Function

def length (*args):
  print(args)
  print(type(args))

# Demonstrate the function call outside its definition
length(5,3,5,7,3,8,2)
```

```python
# Create an Arbitrary Argument Function to find a sum of list

def sum_of_list(*args):
  sum = 0
  for num in args:
    sum += num
  return sum

print("Sum:", sum_of_list(1, 2, 3, 4, 5))
print("Sum:", sum_of_list())
```
