```python
# creat a user define fun just like len fun do not use len

def len():
  a=input("Enter the String: ")
  count=0
  for i in a:
    count=count+1
  return count
print(len())
```

```python
def large(num_list):
  max_element = num_list[0]
  for i in num_list:
    if i > max_element:
      max_element = i
  return max_element

a_str = input("Enter a list of numbers separated by commas (e.g., 1,2,3): ")
# Split the string by commas and convert each part to an integer
a = [int(i.strip()) for i in a_str.split(',') if i.strip()]

# Check if the list is not empty before calling large()
if a:
  print("The maximum element is:", large(a))
else:
  print("No numbers entered.")
```
