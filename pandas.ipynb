```python
import pandas as pd
s=pd.Series([2,4,7,9,10,1,3,5,6,34,67,78,34,23,36,27])
print(type(s))
print(s)
```

```python
df=pd.DataFrame({
    "Roll No":[2,4,7,9,4,21,3,5,6,34],
    "Name":["Avi","Rahul","Ram","Sham","Rahul","Yash","Sakshi","Pooja","Raja","Abhi"],
    "Class":["A","B","C","D","B","F","G","H","None","J"],
    "Age":[21,22,None,24,24,25,21,27,28,29],
    "Marks":[80,60,None,64,62,52,None,42,67,52]
})
print(type(df))
print(df)
```
